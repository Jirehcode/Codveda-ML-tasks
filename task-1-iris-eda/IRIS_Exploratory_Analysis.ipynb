```python
# importing required libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Loading the CSV file into pandas dataframe
df = pd.read_csv("/content/drive/MyDrive/Codveda Data science Task List/1) iris.csv")
df.head(5)
```

```python
# Finding the null values
null_values = df.isnull().sum()
print(null_values)
```

```python
# Finding the mean
mean_value = df.mean(numeric_only=True)
print(mean_value)
```

```python
# Finding the median
median_value = df.median(numeric_only=True)
print(median_value)
```

```python
# Finding the mode
mode_value = df.mode(numeric_only=True)
print(mode_value)
```

```python
# Finding the variance
variance_value = df.var(numeric_only=True)
print(variance_value)
```

```python
# Finding the standard deviation
std_value = df.std(numeric_only=True)
print(std_value)
```

```python
# Finding the first quartile
first_quartile = df.quantile(0.25, numeric_only=True)
print(first_quartile)
```

```python
# Finding the second quartile
second_quartile = df.quantile(0.50, numeric_only=True)
print(second_quartile)
```

```python
# Finding the third quartile
third_quartile = df.quantile(0.75, numeric_only=True)
print(third_quartile)
```

```python
# Finding the interquartile range
IQR = third_quartile - first_quartile
print(IQR)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("iris")
plt.figure(figsize=(8,6))

# Define colors for the species
colors = ['#0077BE', 'orange', '#228B22']

# create scatter plot
sns.scatterplot(data=df, x="sepal_length", y="sepal_width", hue="species", style="species", s=70, palette=colors)
# Add title and labels
plt.title("Sepal Length vs Sepal Width", fontsize=14, fontweight="bold")
plt.xlabel("Sepal Length (cm)", fontsize=12 )
plt.ylabel("Sepal Width (cm)", fontsize=12)

plt.tight_layout()
plt.show
```

```python
# Detecting outliers in each Iris attribute based on species
import matplotlib.pyplot as plt
import seaborn as sns
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
colors = ['#0077BE', 'orange', '#228B22', 'red']
for i, feature in enumerate(features):
    sns.boxplot(x='species', y=feature, data=df, color=colors[i], ax=axes[i//2, i%2])
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset("iris")
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
colors = ['#0077BE', 'orange', '#228B22'] # Reduced to 3 colors to match the 3 species
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))
for i, feature in enumerate(features):
  ax =axes[i // 2, i % 2]
  sns.histplot(data=df, x=feature, hue='species', multiple='stack', palette=colors,element="bars",kde=True, ax=ax)
  ax.set_title(f'Histogram of {feature.replace("_",  " ").title()} by species',fontsize=12, fontweight="bold")
  ax.set_xlabel(f'{feature.replace("_", " ").title()} (cm)', fontsize=12)
  ax.set_ylabel('Count', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
# calculate correlation matrix
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
# create heatmap
plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title("Correlation Matrix Heatmap", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```
